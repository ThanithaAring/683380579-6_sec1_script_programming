# **Lab 6.1: Simple Calculator Functions**
○ Objective: Students will define and use functions to perform basic arithmetic
operations, practicing parameter passing and return values.

○ Tasks (Common for Local &amp; Colab):
1. Create a new Python file (lab6_1.py) or Colab code cell.
2. Define four functions: add(a, b), subtract(a, b), multiply(a, b), divide(a, b).
* Each function should take two numbers as parameters.
* ach function should return the result of the operation.
* Add a docstring to each function explaining its purpose.
3. Implement a simple menu-driven program (using a while loop from Week 3) that
allows the user to choose an operation.
4. Prompt the user for two numbers.
5. Call the appropriate function based on user choice and print the returned result.
6. Handle division by zero in the divide function (e.g., return &quot;Error: Division by zero&quot;
or a specific value).
7. Challenge: Add a function power(base, exponent) using default parameters (e.g.,
exponent=2 for squaring if not provided).

```python
# lab6_1.py
"""Lab 6.1: Simple Calculator Functions (menu-driven)."""


def add(a, b):
    """Return the sum of a and b."""
    return a + b


def subtract(a, b):
    """Return the difference of a minus b."""
    return a - b


def multiply(a, b):
    """Return the product of a and b."""
    return a * b


def divide(a, b):
    """Return a divided by b.

    If b is zero, return an error message instead of raising an exception.
    """
    if b == 0:
        return "Error: Division by zero"
    return a / b


def power(base, exponent=2):
    """Return base raised to exponent.

    Challenge: exponent has a default value of 2, so power(x) squares x.
    """
    return base ** exponent


def get_number(prompt):
    """Ask the user for a number repeatedly until a valid number is entered."""
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Invalid input. Please enter a number.")


def show_menu():
    """Print the calculator menu."""
    print("\n===== Simple Calculator =====")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power (default exponent = 2)")
    print("6. Exit")


def main():
    """Run the menu-driven calculator using a while loop."""
    operations = {
        "1": ("+", add),
        "2": ("-", subtract),
        "3": ("*", multiply),
        "4": ("/", divide),
    }

    while True:
        show_menu()
        choice = input("Choose an option (1-6): ").strip()

        if choice == "6":
            print("Goodbye!")
            break

        if choice in operations:
            a = get_number("Enter first number: ")
            b = get_number("Enter second number: ")
            symbol, func = operations[choice]
            result = func(a, b)          # call the function and keep the returned value
            print(f"Result: {a} {symbol} {b} = {result}")

        elif choice == "5":
            base = get_number("Enter base: ")
            exp_text = input("Enter exponent (press Enter to use default 2): ").strip()
            if exp_text == "":
                result = power(base)                     # uses default exponent=2
                print(f"Result: {base} ** 2 = {result}")
            else:
                try:
                    exponent = float(exp_text)
                    result = power(base, exponent)
                    print(f"Result: {base} ** {exponent} = {result}")
                except ValueError:
                    print("Invalid exponent.")
        else:
            print("Invalid choice. Please choose 1-6.")


if __name__ == "__main__":
    main()
```

# **Lab 6.2: Custom Module and Standard Library Usage**
○ Objective: Students will create their own Python module and import it into another

○ Tasks (Common for Local &amp; Colab):

2. Part 2: Use the Custom Module and Standard Modules:
* Create another new Python file (main_program.py) or Colab code cell.

* Import standard modules: import math, import random.
* Call my_utils.greet(&quot;Alice&quot;).
* Test my_utils.is_prime() with a few numbers (e.g., 7, 10).
* Use math.sqrt() to calculate the square root of a number.
* Use random.randint() to generate a random number between 1 and 100.
* For Colab specifics: If my_utils.py is not easily managed as a separate file,
the main_program cell can simply define the functions from my_utils within
itself or the instructor can pre-upload my_utils.py to Colab. The best way for
true module demo in Colab is to use %%writefile my_utils.py magic command
in one cell to create the file, then import it in another.

```python
%%writefile my_utils.py
"""Lab 6.2 Part 1: Custom utility module."""


def greet(name):
    """Print a greeting message for the given name."""
    print(f"Hello, {name}!")


def is_prime(number):
    """Return True if number is a prime number, otherwise False."""
    if number < 2:
        return False
    if number == 2:
        return True
    if number % 2 == 0:
        return False
    i = 3
    while i * i <= number:
        if number % i == 0:
            return False
        i += 2
    return True
```

```python
"""Lab 6.2 Part 2: Use the custom module and standard library modules."""

import my_utils
import math
import random

my_utils.greet("Alice")

for n in [7, 10, 2, 1, 29]:
    print(f"is_prime({n}) = {my_utils.is_prime(n)}")

number = 49
print(f"Square root of {number} = {math.sqrt(number)}")

random_number = random.randint(1, 100)
print(f"Random number between 1 and 100: {random_number}")
print(f"Is {random_number} prime? {my_utils.is_prime(random_number)}")
```

# **Activity 6.1: Refactoring a Previous Lab Assignment**
○ Format: Individual coding.

○ Instructions: Choose one of your completed lab assignments from Week 2
(Conditionals) or Week 3 (Loops). Identify repetitive blocks of code or logical units that

○ Example (Week 2): The &quot;Number Classifier&quot; lab could have functions like

○ Purpose: Practical application of modular design principles, reinforcing function
concepts.

```python
"""Activity 6.1: Refactor the Week 2 "Number Classifier" lab using functions.

---------------- BEFORE (Week 2 version, no functions) ----------------
num = int(input("Enter a number: "))
if num > 0:
    print(f"{num} is positive")
elif num < 0:
    print(f"{num} is negative")
else:
    print(f"{num} is zero")
if num % 2 == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
# (to classify another number, the whole block had to be copied again)
-----------------------------------------------------------------------
"""


def is_positive_negative_zero(num):
    """Return 'positive', 'negative' or 'zero' for the given number."""
    if num > 0:
        return "positive"
    elif num < 0:
        return "negative"
    return "zero"


def is_even_odd(num):
    """Return 'even' or 'odd' for the given integer."""
    return "even" if num % 2 == 0 else "odd"


def classify_number(num):
    """Return a full description string for the given integer."""
    return f"{num} is {is_positive_negative_zero(num)} and {is_even_odd(num)}"


def get_integer(prompt):
    """Ask the user for an integer until a valid one is entered."""
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Please enter a whole number.")


def main():
    """Classify numbers repeatedly until the user types 'q'."""
    while True:
        text = input("\nEnter a number (or 'q' to quit): ").strip()
        if text.lower() == "q":
            print("Goodbye!")
            break
        try:
            num = int(text)
        except ValueError:
            print("Please enter a whole number.")
            continue
        print(classify_number(num))


if __name__ == "__main__":
    main()
```

## 1. ประโยชน์ของ Modular Programming

Modular Programming คือการแบ่งโปรแกรมใหญ่ออกเป็นชิ้นเล็ก ๆ (ฟังก์ชันหรือไฟล์) ให้แต่ละชิ้นทำงานแค่อย่างเดียว เหมือนแบ่งงานบ้านเป็นงานล้างจาน งานกวาดบ้าน งานซักผ้า แทนที่จะทำทุกอย่างปนกัน

ข้อดีข้อแรกคือโค้ดอ่านง่ายขึ้น โปรแกรมใหญ่ ๆ ถ้าแบ่งเป็นไฟล์ย่อย เช่น `students.py` กับ `grades.py` ก็จะรู้ทันทีว่าโค้ดส่วนไหนอยู่ตรงไหน

ข้อดีข้อที่สองคือทำงานเป็นทีมได้ง่าย แต่ละคนเขียนคนละไฟล์ จึงไม่แก้โค้ดทับกัน

ข้อดีข้อที่สามคือหาบั๊กง่าย ถ้าผลหารผิด ก็ไปดูแค่ฟังก์ชัน `divide()` ไม่ต้องอ่านทั้งโปรแกรม

ข้อดีข้อที่สี่คือทดสอบง่าย ลองเรียกฟังก์ชันทีละตัวดูว่าได้ผลถูกไหม เช่น `is_prime(7)` ต้องได้ `True`

ข้อดีข้อสุดท้ายคือใช้ซ้ำได้ เขียนฟังก์ชันครั้งเดียว แล้วเอาไปใช้ในงานอื่นได้อีก เช่น `my_utils.is_prime()`

สรุปง่าย ๆ คือ แบ่งโค้ดเป็นชิ้นเล็ก แล้วจะอ่านง่าย แก้ง่าย และใช้ซ้ำได้

## 2. สำรวจ Python Standard Library

Standard Library คือชุดเครื่องมือที่ Python มีมาให้อยู่แล้ว แค่ `import` ก็ใช้ได้เลย ไม่ต้องเขียนเอง (อ้างอิง: https://docs.python.org/3/library/)

`datetime` ใช้เรื่องวันที่และเวลา เช่น นับว่าเหลืออีกกี่วันถึงวันสอบ

`os` ใช้จัดการไฟล์และโฟลเดอร์ในเครื่อง เช่น ดูว่าในโฟลเดอร์มีไฟล์อะไรบ้าง

`collections` มีตัวช่วยอย่าง `Counter` ไว้นับของ เช่น นับว่ามีคนได้เกรด A กี่คน B กี่คน
